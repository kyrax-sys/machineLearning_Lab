```python

import pandas as pd

url = 'https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv'
df = pd.read_csv(url)

survival_counts = df['Survived'].value_counts()
print("survival count:")
print(survival_counts)

survived_ages = df[df['Survived'] == 1]['Age']
non_survived_ages = df[df['Survived'] == 0]['Age']
print(f"\nAverage age of survivors: {survived_ages.mean():.2f}")
print(f"Average age of non-survivors: {non_survived_ages.mean():.2f}")

elderly_passengers = df[df['Age'] > 60]
print(f"\nPassengers older than 60 years (Total: {len(elderly_passengers)}):")
print(elderly_passengers[['Name', 'Age', 'Survived']].head())

avg_fare = df.groupby('Pclass')['Fare'].mean()
print("\nAverage fare per passenger class:")
print(avg_fare)
```

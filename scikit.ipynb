```python
from sklearn.datasets import load_iris
iris =load_iris()
x=iris.data
y=iris.target
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
from sklearn.tree import DecisionTreeClassifier
dtree=DecisionTreeClassifier()
dtree.fit(x_train,y_train)
```

```python
pred=dtree.predict(x_test)
print(pred)
```

```python
from sklearn.metrics import accuracy_score,precision_score,recall_score,f1_score,confusion_matrix
print("Accuracy Score:",accuracy_score(y_test,pred))
print("Precision Score:",precision_score(y_test,pred,average='weighted'))
print("Recall Score:",recall_score(y_test,pred,average='weighted'))
print("F1 Score:",f1_score(y_test,pred,average='weighted'))

#conmfusion matrix
print("Confusion Matrix:\n",confusion_matrix(y_test,pred))
```

```python
from sklearn.model_selection import cross_val_score
cv_score=cross_val_score(dtree,x,y,cv=5,scoring='accuracy')
print("cv score: ",cv_score)
print("Mean Accuracy:",cv_score.mean())
```

```python
#breast cancer prediction assignment 01
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix


breast_cancer = load_breast_cancer()
x_bc = breast_cancer.data
y_bc = breast_cancer.target

```

```python
# training and testing sets
x_train_bc, x_test_bc, y_train_bc, y_test_bc = train_test_split(x_bc, y_bc, test_size=0.2, random_state=42)

```

```python
# Load and train Logistic Regression model
log_reg_bc = LogisticRegression()
log_reg_bc.fit(x_train_bc, y_train_bc)

```

```python
# Making predictions
pred_bc = log_reg_bc.predict(x_test_bc)
print("Predictions on Breast Cancer test set:", pred_bc)
```

```python
# Calculate and printing classification metrics
print("Accuracy Score:", accuracy_score(y_test_bc, pred_bc))
print("Precision Score:", precision_score(y_test_bc, pred_bc, average='weighted'))
print("Recall Score:", recall_score(y_test_bc, pred_bc, average='weighted'))
print("F1 Score:", f1_score(y_test_bc, pred_bc, average='weighted'))

# Confusion Matrix
print("\nConfusion Matrix:\n", confusion_matrix(y_test_bc, pred_bc))
```

```python
#wine dataset classificiation using knn assignment 02
from sklearn.datasets import load_wine
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix

# Loading dataset
wine = load_wine()
x_wine = wine.data
y_wine = wine.target
```

```python
# Spliting into training and testing set
x_train_wine, x_test_wine, y_train_wine, y_test_wine = train_test_split(x_wine, y_wine, test_size=0.2, random_state=42)

```

```python
# Load and train knn model

knn_wine = KNeighborsClassifier()
knn_wine.fit(x_train_wine, y_train_wine)

```

```python
# Making predictions
pred_wine = knn_wine.predict(x_test_wine)
print("Predictions on Wine test set:", pred_wine)
```

```python
# Calculating and printing classification metrics
print("Accuracy Score:", accuracy_score(y_test_wine, pred_wine))
print("Precision Score:", precision_score(y_test_wine, pred_wine, average='weighted'))
print("Recall Score:", recall_score(y_test_wine, pred_wine, average='weighted'))
print("F1 Score:", f1_score(y_test_wine, pred_wine, average='weighted'))

# Confusion Matrix
print("\nConfusion Matrix:\n", confusion_matrix(y_test_wine, pred_wine))
```

```python
#diabetic dataset regression using linear regression
from sklearn.datasets import load_diabetes
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# Load the dataset
diabetes = load_diabetes()
x_diabetes = diabetes.data
y_diabetes = diabetes.target
```

```python
# Spliting into training and testing set
x_train_diabetes, x_test_diabetes, y_train_diabetes, y_test_diabetes = train_test_split(x_diabetes, y_diabetes, test_size=0.2, random_state=42)

```

```python
# Load and train Linear Regression model
linear_reg_diabetes = LinearRegression()
linear_reg_diabetes.fit(x_train_diabetes, y_train_diabetes)

```

```python
# Making predictions
pred_diabetes = linear_reg_diabetes.predict(x_test_diabetes)
print(pred_diabetes)
```

```python
# Calculate and print regression metrics

print("Mean Absolute Error (MAE):", mean_absolute_error(y_test_diabetes, pred_diabetes))
print("Mean Squared Error (MSE):", mean_squared_error(y_test_diabetes, pred_diabetes))
print("R-squared (R2) Score:", r2_score(y_test_diabetes, pred_diabetes))
```

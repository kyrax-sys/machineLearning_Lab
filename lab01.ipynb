```python
#activity2_nump_manipal_temp
days = np.arange(1, 32)

temps = np.array([
    27, 27, 28, 28, 28, 28, 28,
    28, 28, 28, 28, 29, 28, 28,
    28, 28, 27, 28, 28, 28, 28,
    28, 29, 29, 29, 30, 29, 29,
    28, 28, 29
])
```

```python
hot = np.max(temps)
cold = np.min(temps)

print(f"hottest temp: {hot}°C")
print(f"coldest temp: {cold}°C")
```

```python

heat = temps >= 35
hot_days = days[heat]
print("days with temp >= 35°C:", hot_days if len(hot_days) else "not found")

print(f"average temperature: {np.mean(temps)}°C")
```
